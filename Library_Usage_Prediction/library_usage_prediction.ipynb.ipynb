```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve
)
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 90)
print("23. EDUCATION - PREDICTING LIBRARY USAGE")
print("=" * 90)

FILE_PATH = "dataset_univLib.csv"

df = pd.read_csv(
    FILE_PATH,
    encoding="utf-8-sig",
    sep=None,
    engine="python",
    on_bad_lines="skip"
)

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace("-", "_", regex=False)
)

print("\n1. DATASET LOADED")
print("-" * 90)
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nFirst 5 Records:")
display(df.head())

print("\n2. DATASET INFORMATION")
print("-" * 90)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()

print("\n3. DATA CLEANING")
print("-" * 90)

duplicate_count = df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)
print("Missing Values Before Cleaning:", df.isnull().sum().sum())

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)

numeric_columns = [
    "membership_number",
    "book_id_number",
    "borrow_duration_(days)",
    "cumulative_gpa",
    "total_credit_cumulative",
    "academic_year",
    "location_code",
    "borrowed"
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

date_columns = [
    "borrow_date",
    "return_date",
    "date_of_birth"
]

for column in date_columns:
    if column in df.columns:
        df[column] = pd.to_datetime(
            df[column],
            errors="coerce"
        )

categorical_columns = [
    "book_title",
    "department_code",
    "gender",
    "place_of_birth",
    "program_code",
    "isbn",
    "ddc_number",
    "author",
    "published_year",
    "publisher_name",
    "location_code"
]

for column in categorical_columns:
    if column in df.columns:
        df[column] = df[column].astype("string").str.strip()

for column in df.select_dtypes(include=np.number).columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].median()
        )

for column in df.select_dtypes(
    include=["object", "string", "category"]
).columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()
        if len(mode_value) > 0:
            df[column] = df[column].fillna(
                mode_value.iloc[0]
            )
        else:
            df[column] = df[column].fillna("Unknown")

print("Missing Values After Cleaning:", df.isnull().sum().sum())
print("Duplicate Rows After Cleaning:", df.duplicated().sum())
print("Final Transaction Dataset Shape:", df.shape)

print("\n4. DATE FEATURE ENGINEERING")
print("-" * 90)

if "borrow_date" in df.columns:

    df["borrow_year"] = df["borrow_date"].dt.year
    df["borrow_month"] = df["borrow_date"].dt.month
    df["borrow_month_name"] = df["borrow_date"].dt.month_name()
    df["borrow_day"] = df["borrow_date"].dt.day
    df["borrow_day_of_week"] = df["borrow_date"].dt.dayofweek

if "return_date" in df.columns and "borrow_date" in df.columns:

    df["actual_borrow_duration"] = (
        df["return_date"] - df["borrow_date"]
    ).dt.days

print("Date features created successfully.")

print("\n5. BORROWING DATA ANALYSIS")
print("-" * 90)

if "borrowed" in df.columns:

    print("Borrowed Distribution:")
    display(
        df["borrowed"]
        .value_counts()
        .to_frame("Transactions")
    )

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="borrowed"
    )

    plt.title("Borrowing Status Distribution")
    plt.xlabel("Borrowed")
    plt.ylabel("Number of Transactions")

    plt.tight_layout()
    plt.show()

print("\n6. STUDENT-LEVEL DATASET")
print("-" * 90)

if "membership_number" not in df.columns:
    raise ValueError(
        "Membership Number was not detected after cleaning."
    )

student_df = (
    df.groupby("membership_number")
    .agg(
        total_transactions=(
            "membership_number",
            "size"
        ),
        total_books_borrowed=(
            "book_id_number",
            "nunique"
        ),
        average_borrow_duration=(
            "borrow_duration_(days)",
            "mean"
        ),
        total_borrow_duration=(
            "borrow_duration_(days)",
            "sum"
        ),
        average_gpa=(
            "cumulative_gpa",
            "mean"
        ),
        total_credits=(
            "total_credit_cumulative",
            "max"
        ),
        gender=(
            "gender",
            "first"
        ),
        place_of_birth=(
            "place_of_birth",
            "first"
        ),
        academic_year=(
            "academic_year",
            "first"
        ),
        program_code=(
            "program_code",
            "first"
        ),
        department_code=(
            "department_code",
            "first"
        ),
        unique_authors=(
            "author",
            "nunique"
        ),
        unique_publishers=(
            "publisher_name",
            "nunique"
        ),
        unique_subjects=(
            "ddc_number",
            "nunique"
        ),
        first_borrow_date=(
            "borrow_date",
            "min"
        ),
        last_borrow_date=(
            "borrow_date",
            "max"
        ),
        total_borrowed=(
            "borrowed",
            "sum"
        )
    )
    .reset_index()
)

print("Student-Level Dataset Shape:")
print(student_df.shape)

print("\nStudent-Level Dataset:")
display(student_df.head())

print("\n7. CREATE TARGET VARIABLE")
print("-" * 90)

threshold = student_df[
    "total_transactions"
].quantile(0.75)

print(
    "75th Percentile Transaction Threshold:",
    threshold
)

student_df["frequent_library_user"] = (
    student_df["total_transactions"]
    >= threshold
).astype(int)

if student_df[
    "frequent_library_user"
].nunique() < 2:

    threshold = student_df[
        "total_transactions"
    ].median()

    student_df["frequent_library_user"] = (
        student_df["total_transactions"]
        >= threshold
    ).astype(int)

print("\nTarget Distribution:")

target_distribution = (
    student_df["frequent_library_user"]
    .value_counts()
    .rename({
        0: "Non-Frequent User",
        1: "Frequent User"
    })
)

display(
    target_distribution.to_frame("Students")
)

print("\nTarget Percentage:")

target_percentage = (
    student_df["frequent_library_user"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .rename({
        0: "Non-Frequent User",
        1: "Frequent User"
    })
)

display(
    target_percentage.to_frame("Percentage")
)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=student_df,
    x="frequent_library_user"
)

plt.title("Frequent vs Non-Frequent Library Users")
plt.xlabel("Library User Category")
plt.ylabel("Number of Students")

plt.xticks(
    [0, 1],
    ["Non-Frequent User", "Frequent User"]
)

plt.tight_layout()
plt.show()

print("\n8. EXPLORATORY DATA ANALYSIS")
print("-" * 90)

eda_columns = [
    "total_transactions",
    "total_books_borrowed",
    "average_borrow_duration",
    "average_gpa",
    "total_credits",
    "unique_authors",
    "unique_publishers",
    "unique_subjects"
]

for column in eda_columns:

    if column in student_df.columns:

        plt.figure(figsize=(8, 5))

        sns.histplot(
            data=student_df,
            x=column,
            kde=True
        )

        plt.title(
            f"Distribution of {column.replace('_', ' ').title()}"
        )

        plt.xlabel(
            column.replace("_", " ").title()
        )

        plt.ylabel("Frequency")

        plt.tight_layout()
        plt.show()

print("\n9. CORRELATION HEATMAP")
print("-" * 90)

correlation_columns = student_df.select_dtypes(
    include=np.number
).columns

plt.figure(figsize=(14, 10))

sns.heatmap(
    student_df[
        correlation_columns
    ].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()

print("\n10. LIBRARY USAGE BY ACADEMIC PROGRAM")
print("-" * 90)

program_analysis = (
    student_df
    .groupby("program_code")
    .agg(
        students=("membership_number", "count"),
        average_transactions=(
            "total_transactions",
            "mean"
        ),
        average_books=(
            "total_books_borrowed",
            "mean"
        ),
        frequent_users=(
            "frequent_library_user",
            "sum"
        ),
        frequent_user_rate=(
            "frequent_library_user",
            "mean"
        )
    )
    .sort_values(
        "frequent_user_rate",
        ascending=False
    )
)

program_analysis[
    "frequent_user_rate"
] *= 100

display(program_analysis)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=student_df,
    x="program_code",
    y="frequent_library_user",
    estimator=np.mean
)

plt.title(
    "Frequent Library Usage by Academic Program"
)

plt.xlabel("Program")
plt.ylabel("Frequent User Rate")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

print("\n11. LIBRARY USAGE BY ACADEMIC YEAR")
print("-" * 90)

year_analysis = (
    student_df
    .groupby("academic_year")
    .agg(
        students=("membership_number", "count"),
        average_transactions=(
            "total_transactions",
            "mean"
        ),
        frequent_users=(
            "frequent_library_user",
            "sum"
        ),
        frequent_user_rate=(
            "frequent_library_user",
            "mean"
        )
    )
)

year_analysis[
    "frequent_user_rate"
] *= 100

display(year_analysis)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=student_df,
    x="academic_year",
    y="frequent_library_user",
    estimator=np.mean
)

plt.title(
    "Frequent Library Usage by Academic Year"
)

plt.xlabel("Academic Year")
plt.ylabel("Frequent User Rate")

plt.tight_layout()
plt.show()

print("\n12. LIBRARY USAGE BY GENDER")
print("-" * 90)

gender_analysis = (
    student_df
    .groupby("gender")
    .agg(
        students=("membership_number", "count"),
        average_transactions=(
            "total_transactions",
            "mean"
        ),
        frequent_users=(
            "frequent_library_user",
            "sum"
        ),
        frequent_user_rate=(
            "frequent_library_user",
            "mean"
        )
    )
)

gender_analysis[
    "frequent_user_rate"
] *= 100

display(gender_analysis)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=student_df,
    x="gender",
    y="frequent_library_user",
    estimator=np.mean
)

plt.title("Frequent Library Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Frequent User Rate")

plt.tight_layout()
plt.show()

print("\n13. BORROWING DURATION ANALYSIS")
print("-" * 90)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=student_df,
    x="frequent_library_user",
    y="average_borrow_duration"
)

plt.title(
    "Borrowing Duration by Library User Category"
)

plt.xlabel("Library User Category")
plt.ylabel("Average Borrow Duration")

plt.xticks(
    [0, 1],
    ["Non-Frequent", "Frequent"]
)

plt.tight_layout()
plt.show()

print("\n14. BOOK BORROWING ANALYSIS")
print("-" * 90)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=student_df,
    x="frequent_library_user",
    y="total_books_borrowed"
)

plt.title(
    "Books Borrowed by Library User Category"
)

plt.xlabel("Library User Category")
plt.ylabel("Number of Unique Books")

plt.xticks(
    [0, 1],
    ["Non-Frequent", "Frequent"]
)

plt.tight_layout()
plt.show()

print("\n15. GPA ANALYSIS")
print("-" * 90)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=student_df,
    x="frequent_library_user",
    y="average_gpa"
)

plt.title(
    "GPA by Library User Category"
)

plt.xlabel("Library User Category")
plt.ylabel("Average GPA")

plt.xticks(
    [0, 1],
    ["Non-Frequent", "Frequent"]
)

plt.tight_layout()
plt.show()

print("\n16. BORROWING TREND BY YEAR")
print("-" * 90)

if "borrow_year" in df.columns:

    yearly_usage = (
        df.groupby("borrow_year")
        .size()
        .reset_index(
            name="transactions"
        )
    )

    display(yearly_usage)

    plt.figure(figsize=(10, 6))

    sns.lineplot(
        data=yearly_usage,
        x="borrow_year",
        y="transactions",
        marker="o"
    )

    plt.title(
        "Library Borrowing Transactions by Year"
    )

    plt.xlabel("Year")
    plt.ylabel("Transactions")

    plt.tight_layout()
    plt.show()

print("\n17. BORROWING TREND BY MONTH")
print("-" * 90)

if "borrow_month_name" in df.columns:

    monthly_usage = (
        df.groupby(
            "borrow_month_name"
        )
        .size()
        .reset_index(
            name="transactions"
        )
    )

    month_order = [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July",
        "August",
        "September",
        "October",
        "November",
        "December"
    ]

    monthly_usage[
        "month_number"
    ] = monthly_usage[
        "borrow_month_name"
    ].map({
        month: i
        for i, month in enumerate(
            month_order,
            1
        )
    })

    monthly_usage = (
        monthly_usage
        .sort_values("month_number")
    )

    display(monthly_usage)

    plt.figure(figsize=(12, 6))

    sns.barplot(
        data=monthly_usage,
        x="borrow_month_name",
        y="transactions"
    )

    plt.title(
        "Library Borrowing Transactions by Month"
    )

    plt.xlabel("Month")
    plt.ylabel("Transactions")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

print("\n18. MACHINE LEARNING PREPARATION")
print("-" * 90)

target = "frequent_library_user"

X = student_df.drop(
    columns=[
        "frequent_library_user",
        "membership_number",
        "total_transactions"
    ]
)

y = student_df[target]

if "first_borrow_date" in X.columns:
    X["first_borrow_date"] = (
        X["first_borrow_date"]
        .astype("int64")
    )

if "last_borrow_date" in X.columns:
    X["last_borrow_date"] = (
        X["last_borrow_date"]
        .astype("int64")
    )

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

print("Features:")
print(X.columns.tolist())

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

print("\n19. TRAIN TEST SPLIT")
print("-" * 90)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))

print("\n20. PREPROCESSING")
print("-" * 90)

try:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            encoder
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)

print("Preprocessing completed.")

print("\n21. DECISION TREE")
print("-" * 90)

decision_tree = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            DecisionTreeClassifier(
                criterion="gini",
                max_depth=6,
                min_samples_split=10,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

decision_tree.fit(
    X_train,
    y_train
)

dt_pred = decision_tree.predict(
    X_test
)

dt_prob = decision_tree.predict_proba(
    X_test
)[:, 1]

dt_accuracy = accuracy_score(
    y_test,
    dt_pred
)

dt_precision = precision_score(
    y_test,
    dt_pred,
    zero_division=0
)

dt_recall = recall_score(
    y_test,
    dt_pred,
    zero_division=0
)

dt_f1 = f1_score(
    y_test,
    dt_pred,
    zero_division=0
)

dt_auc = roc_auc_score(
    y_test,
    dt_prob
)

print("Accuracy :", round(dt_accuracy, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1 Score :", round(dt_f1, 4))
print("ROC-AUC  :", round(dt_auc, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        dt_pred,
        target_names=[
            "Non-Frequent",
            "Frequent"
        ],
        zero_division=0
    )
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    confusion_matrix(
        y_test,
        dt_pred
    ),
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Non-Frequent",
        "Frequent"
    ],
    yticklabels=[
        "Non-Frequent",
        "Frequent"
    ]
)

plt.title(
    "Decision Tree Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()

print("\n22. SVM")
print("-" * 90)

svm = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            SVC(
                kernel="rbf",
                C=1.0,
                gamma="scale",
                probability=True,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

svm.fit(
    X_train,
    y_train
)

svm_pred = svm.predict(
    X_test
)

svm_prob = svm.predict_proba(
    X_test
)[:, 1]

svm_accuracy = accuracy_score(
    y_test,
    svm_pred
)

svm_precision = precision_score(
    y_test,
    svm_pred,
    zero_division=0
)

svm_recall = recall_score(
    y_test,
    svm_pred,
    zero_division=0
)

svm_f1 = f1_score(
    y_test,
    svm_pred,
    zero_division=0
)

svm_auc = roc_auc_score(
    y_test,
    svm_prob
)

print("Accuracy :", round(svm_accuracy, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("ROC-AUC  :", round(svm_auc, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        svm_pred,
        target_names=[
            "Non-Frequent",
            "Frequent"
        ],
        zero_division=0
    )
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    confusion_matrix(
        y_test,
        svm_pred
    ),
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=[
        "Non-Frequent",
        "Frequent"
    ],
    yticklabels=[
        "Non-Frequent",
        "Frequent"
    ]
)

plt.title(
    "SVM Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()

print("\n23. MODEL COMPARISON")
print("-" * 90)

comparison = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "SVM"
    ],
    "Accuracy": [
        dt_accuracy,
        svm_accuracy
    ],
    "Precision": [
        dt_precision,
        svm_precision
    ],
    "Recall": [
        dt_recall,
        svm_recall
    ],
    "F1-Score": [
        dt_f1,
        svm_f1
    ],
    "ROC-AUC": [
        dt_auc,
        svm_auc
    ]
})

display(
    comparison.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC"
]

x = np.arange(
    len(metrics)
)

width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    x - width / 2,
    comparison.loc[0, metrics],
    width,
    label="Decision Tree"
)

plt.bar(
    x + width / 2,
    comparison.loc[1, metrics],
    width,
    label="SVM"
)

plt.xticks(
    x,
    metrics
)

plt.ylim(
    0,
    1.05
)

plt.ylabel("Score")
plt.title(
    "Decision Tree vs SVM Performance"
)

plt.legend()

plt.tight_layout()
plt.show()

print("\n24. ROC CURVE")
print("-" * 90)

dt_fpr, dt_tpr, _ = roc_curve(
    y_test,
    dt_prob
)

svm_fpr, svm_tpr, _ = roc_curve(
    y_test,
    svm_prob
)

plt.figure(figsize=(9, 6))

plt.plot(
    dt_fpr,
    dt_tpr,
    label=f"Decision Tree AUC = {dt_auc:.3f}"
)

plt.plot(
    svm_fpr,
    svm_tpr,
    label=f"SVM AUC = {svm_auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve - Library Usage Prediction"
)

plt.legend()

plt.tight_layout()
plt.show()

print("\n25. DECISION TREE FEATURE IMPORTANCE")
print("-" * 90)

trained_preprocessor = (
    decision_tree
    .named_steps["preprocessor"]
)

trained_tree = (
    decision_tree
    .named_steps["classifier"]
)

feature_names = (
    trained_preprocessor
    .get_feature_names_out()
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": trained_tree.feature_importances_
})

feature_importance = (
    feature_importance
    .sort_values(
        "Importance",
        ascending=False
    )
)

display(
    feature_importance.head(20)
)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=feature_importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title(
    "Top Factors Associated with Frequent Library Usage"
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()

print("\n26. PERMUTATION IMPORTANCE")
print("-" * 90)

permutation = permutation_importance(
    decision_tree,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="f1"
)

permutation_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance_Mean": permutation.importances_mean,
    "Importance_STD": permutation.importances_std
})

permutation_df = (
    permutation_df
    .sort_values(
        "Importance_Mean",
        ascending=False
    )
)

display(permutation_df)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=permutation_df.head(15),
    x="Importance_Mean",
    y="Feature"
)

plt.title(
    "Permutation Importance of Student Characteristics"
)

plt.xlabel("Mean Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()

print("\n27. BEST MODEL")
print("-" * 90)

if svm_f1 > dt_f1:

    best_model = svm
    best_model_name = "SVM"
    best_accuracy = svm_accuracy
    best_precision = svm_precision
    best_recall = svm_recall
    best_f1 = svm_f1
    best_auc = svm_auc

else:

    best_model = decision_tree
    best_model_name = "Decision Tree"
    best_accuracy = dt_accuracy
    best_precision = dt_precision
    best_recall = dt_recall
    best_f1 = dt_f1
    best_auc = dt_auc

print("Best Model:", best_model_name)
print("Accuracy :", round(best_accuracy, 4))
print("Precision:", round(best_precision, 4))
print("Recall   :", round(best_recall, 4))
print("F1 Score :", round(best_f1, 4))
print("ROC-AUC  :", round(best_auc, 4))

print("\n28. SAVE MODEL")
print("-" * 90)

MODEL_FILE = "library_usage_prediction_model.pkl"

joblib.dump(
    best_model,
    MODEL_FILE
)

print(
    "Model saved successfully:",
    MODEL_FILE
)

loaded_model = joblib.load(
    MODEL_FILE
)

verification = loaded_model.predict(
    X_test
)

print(
    "Saved Model Verification Accuracy:",
    round(
        accuracy_score(
            y_test,
            verification
        ),
        4
    )
)

print("\n29. IMPORTANT FACTORS")
print("-" * 90)

print(
    "Top Predictive Factors:"
)

for feature in permutation_df.head(5)["Feature"]:
    print("•", feature)

print("\n30. LIBRARY RESOURCE PLANNING RECOMMENDATIONS")
print("-" * 90)

if len(program_analysis) > 0:

    top_program = (
        program_analysis
        .index[0]
    )

    top_program_rate = (
        program_analysis
        .iloc[0]["frequent_user_rate"]
    )

    print(
        f"1. Program {top_program} has the highest "
        f"frequent-user rate of "
        f"{top_program_rate:.2f}%."
    )

if len(year_analysis) > 0:

    top_year = (
        year_analysis[
            "frequent_user_rate"
        ]
        .idxmax()
    )

    top_year_rate = (
        year_analysis[
            "frequent_user_rate"
        ]
        .max()
    )

    print(
        f"2. Academic year {top_year} has the highest "
        f"frequent-user rate of "
        f"{top_year_rate:.2f}%."
    )

print(
    "3. Increase seating and study-space capacity "
    "during periods of high borrowing activity."
)

print(
    "4. Maintain sufficient copies of frequently "
    "borrowed books."
)

print(
    "5. Allocate library staff based on historical "
    "borrowing demand."
)

print(
    "6. Use program-level usage patterns when "
    "planning library resources."
)

print(
    "7. Monitor popular authors, subjects and "
    "publishers to improve collection planning."
)

print(
    "8. Use historical borrowing behavior to "
    "identify likely frequent users."
)

print(
    "9. Use academic year and program information "
    "to improve resource allocation."
)

print(
    "10. Examination schedules, semester information "
    "and course load are not available in the supplied "
    "dataset and should be added for future analysis."
)

print("\n31. FINAL RESULT")
print("=" * 90)

print(
    "Original Transaction Records:",
    len(df)
)

print(
    "Student Records:",
    len(student_df)
)

print(
    "ML Features:",
    X.shape[1]
)

print(
    "Training Records:",
    len(X_train)
)

print(
    "Testing Records:",
    len(X_test)
)

print("\nDecision Tree")
print("Accuracy :", round(dt_accuracy, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1-Score :", round(dt_f1, 4))
print("ROC-AUC  :", round(dt_auc, 4))

print("\nSVM")
print("Accuracy :", round(svm_accuracy, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1-Score :", round(svm_f1, 4))
print("ROC-AUC  :", round(svm_auc, 4))

print("\nBEST MODEL:", best_model_name)

print("\n" + "=" * 90)
print(
    "LIBRARY USAGE PREDICTION EXPERIMENT COMPLETED SUCCESSFULLY"
)
print("=" * 90)
```
